```python
# Q1. Write a program to calculate the electricity bill
# using only if statement? (accept number of unit from user)
# according to the following criteria:

# Unit Price
# First 100 units no charge
# Next 100 units Rs 5 per unit
# After 200 units Rs 10 per unit
# (For example, if input unit is 350 than total bill amount is Rs2000). Lets explain...
# For first 100 no charges
# Now you have 250 unit remaining
# 100-200 you need to pay – 100*5 = 500
# Now 200-100 you need to pay 150*10 = 1500
# Total = 500+1500 = 2000
# Get input from user
units = int(input("Enter number of units consumed: "))

# Initialize bill
bill = 0

print(f"\nCalculating bill for {units} units:")
print("-" * 30)

# First 100 units - Free
if units > 0:
    if units <= 100:
        print(f"First {units} units: Free")
    if units > 100:
        print("First 100 units: Free")

# Next 100 units (101-200) at Rs 5 per unit
if units > 100:
    if units <= 200:
        # Units between 101-200
        charged_units = units - 100
        cost = charged_units * 5
        bill = bill + cost
        print(f"Next {charged_units} units: {charged_units} × Rs 5 = Rs {cost}")
    
    if units > 200:
        # Full 100 units in this tier
        cost = 100 * 5
        bill = bill + cost
        print(f"Next 100 units (101-200): 100 × Rs 5 = Rs {cost}")

# Above 200 units at Rs 10 per unit
if units > 200:
    remaining_units = units - 200
    cost = remaining_units * 10
    bill = bill + cost
    print(f"Remaining {remaining_units} units (201-{units}): {remaining_units} × Rs 10 = Rs {cost}")

print("-" * 30)
print(f"Total Bill: Rs {bill}")
```

```python
# Q2. Write a program to accept percentage from the user and display
# the grade according to the following criteria:

# Marks Grade
# > 90 A
# > 80 and <= 90 B
# >= 60 and <= 80 C
# below 60 D

percentage = float(input("Enter your percentage: "))

print(f"\nYour percentage: {percentage}%")

# Determine grade using if statements
if percentage > 90:
    grade = "A"
    print("Grade: A")
    print("Excellent! Outstanding performance!")

if percentage > 80 and percentage <= 90:
    grade = "B"
    print("Grade: B") 
    print("Very Good! Keep it up!")

if percentage >= 60 and percentage <= 80:
    grade = "C"
    print("Grade: C")
    print("Good! You can do better!")

if percentage < 60:
    grade = "D"
    print("Grade: D")
    print("Need improvement! Work harder!")

if percentage < 0 or percentage > 100:
    print("Invalid percentage! Should be between 0-100")
```

```python
# Simple if-else code to find second largest number
print("Enter three numbers:")
a = int(input("First number: "))
b = int(input("Second number: "))
c = int(input("Third number: "))

# Find second largest using if-else
if a >= b and a >= c:
    # a is largest
    if b >= c:
        second_largest = b
    else:
        second_largest = c
elif b >= a and b >= c:
    # b is largest
    if a >= c:
        second_largest = a
    else:
        second_largest = c
else:
    # c is largest
    if a >= b:
        second_largest = a
    else:
        second_largest = b

print(f"Second largest number is: {second_largest}")
```

```python
# Accept marked price and calculate net amount after discount
marked_price = float(input("Enter the marked price: "))

# Calculate discount percentage based on marked price
if marked_price > 10000:
    discount_percent = 20
elif marked_price > 7000 and marked_price <= 10000:
    discount_percent = 15
else:  # marked_price <= 7000
    discount_percent = 10

# Calculate discount amount and net amount
discount_amount = (marked_price * discount_percent) / 100
net_amount = marked_price - discount_amount

# Display results
print(f"\nMarked Price: Rs. {marked_price}")
print(f"Discount: {discount_percent}%")
print(f"Discount Amount: Rs. {discount_amount}")
print(f"Net Amount to Pay: Rs. {net_amount}")
```
