```python
#Question 4: Write a Python program to count the number of even and odd numbers from a series of numbers using a for loop. 

user_num = int(input("Enter a number: "))
lst = []
even = 0
odd = 0

for i in range(user_num):
    lst.append(i)

for j in lst:
    if j % 2 == 0:
        even = even + 1
    else:
        odd = odd + 1

print(f"Number of even = {even}")
print(f"Number of odd = {odd}")
    
```

```python
# Question 5: Write a Python program to find the factorial of a number using a for loop.
fact_num=int(input("enter a numer"))
fact=1
for i in range(fact_num,0,-1):
    fact=fact*i
print(f"factorial is {fact}")
```

```python
#Question 6: Write a Python program to print the Fibonacci sequence up to a specified number using a for loop
n = int(input("Enter the number of terms: "))

# Handle edge cases
if n <= 0:
    print("Please enter a positive number")
elif n == 1:
    print("Fibonacci sequence: 0")
elif n == 2:
    print("Fibonacci sequence: 0 1")
else:
    a = 0
    b = 1
    print("Fibonacci sequence:", end=" ")
    print(a, end=" ")  
    print(b, end=" ")  
    
    for i in range(2, n):
        next_term = a + b
        print(next_term, end=" ")
        a = b
        b = next_term
    
    print()  
```

```python
#Question 7: Write a Python program to check if a given number is prime or not using a for loop.
user_num=int(input("enter a  number"))
temp=user_num
count=0
for i in range(2,user_num+1):
    if temp%i==0:
        count=count+1
if count >1:
    print("is not prime")
else:
    print(f" {user_num} is prime")
```

#Question 8: Write a Python program to find the largest element in a list using a for loop. 
n = int(input("Enter the number of elements: "))

numbers = []
print("Enter the elements:")
for i in range(n):
    element = int(input(f"Element {i+1}: "))
    numbers.append(element)


if len(numbers) > 0:
    largest = numbers[0]  # first element is largest
    
    for num in numbers:
        if num > largest:
            largest = num
    
    print(f"List: {numbers}")
    print(f"Largest element: {largest}")
else:
    print("List is empty!")
    

```python
# Question 9: Write a Python program to reverse a given string using a for loop
string2 = input("Enter another string: ")
reversed_string2 = ""

for char in string2:
    reversed_string2 = char + reversed_string2

print(f"Original string: {string2}")
print(f"Reversed string: {reversed_string2}")
```
